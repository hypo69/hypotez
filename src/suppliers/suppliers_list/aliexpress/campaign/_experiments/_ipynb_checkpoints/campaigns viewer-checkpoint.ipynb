```python
import header
import re
from pathlib import Path
from typing import List, Dict, Union, Optional
from types import SimpleNamespace

from src import gs
#from src.suppliers.aliexpress import AliPromoCampaign
from src.suppliers.aliexpress import AliAffiliatedProducts
from src.utils import get_filenames, get_directory_names, read_text_file, csv2dict
from src.utils import list2string
from src.utils import j_loads_ns
from src.utils import pprint
from src.logger import logger

campaigns_directory = Path(gs.path.external_data,'aliexpress','campaigns')
campaign_names = get_directory_names(campaigns_directory)
languages = {'EN': 'USD', 'he': 'ILS', 'ru': 'ILS'}
```

```python

class AliPromoCampaign:
    """! Create Promotional message with products and descriptions inside
    """
    def __init__(self, campaign_name: str, category_name: str, language: str = 'EN', currency: str = 'USD'):
        """! Class initializer for the promotional campaign
        @param campaign_name `str`: The name of the promotional campaign (directory)
        @param category_name `str`: The category of the campaign (directory)
        @param language `str`: The language of the campaign (default 'EN')
        @param currency `str`: The currency used in the campaign (default 'USD')
        @todo: move the path to `event_promotion` directory to JSON settings
        """
        self.campaign_name: str = campaign_name
        self.language: str = language
        self.currency: str = currency
        self.campaign_root: Path = Path(gs.path.external_data, 'aliexpress', 'campaigns', self.campaign_name)
        self.campaign_category_path: Path = self.campaign_root / 'categories' / category_name 
        
        self.campaign: SimpleNamespace = j_loads_ns(self.campaign_root / 'header.json')
        self.category_ns: Optional[SimpleNamespace] = self.get_category(category_name) 
        self.category_name: str = category_name
        self.title: str = f"{self.campaign.title}; {self.category_ns.description if self.category_ns else ''}"
        self.tags: str = list2string(self.campaign.tags)
        
        self.products = self.get_campaign_products()  # Prepare products for the promotional campaign
        if not self.products:
            """! If no products are found, prepare them for the campaign """
            self.products = self.prepare_products_for_campaign()
        if not self.products:
            logger.error(f"Product list is not found!")

    def get_category(self, category_name: str) -> Optional[SimpleNamespace]:
        """! Get the value of the category from JSON campaign
        @param category_name: The category key to look for
        @return: The value of the category if found, else None
        """
        campaign_dict = vars(self.campaign)
        categories = campaign_dict.get('categories', {})

        if isinstance(categories, dict):
            return SimpleNamespace(**categories.get(category_name, {}))
        logger.error(f"'categories' is not a dictionary or '{category_name}' does not exist.")
        return

    def prepare_products_for_campaign(self) -> List[SimpleNamespace]:
        """! Creates a list of products and writes them to JSON files <product_id>.json
        @returns Returns a list of processed products
        """
        campaign_parser = AliAffiliatedProducts(
            campaign_name=self.campaign_name,
            campaign_category=self.category_name,
            language=self.language,
            currency=self.currency
        )

        
        prod_urls = read_text_file(self.campaign_category_path / 'sources.txt') 
        if prod_urls:
            # 1. Вначале пытаюсь получить список из текстового файла собранного из экспорта onetab
            prod_ids_raw = self.extract_prod_ids(prod_urls)
            
            if not prod_ids_raw: # <- вдруг файл пустой
                logger.warning("No product IDs found in the URLs!")
                #return []

            # 2. если не нашел - пытаюсь прочитать файлы из директории sources, которые получил из save link as в почте
            prod_ids_raw = get_filenames(self.campaign_category_path / 'sources')
            if not prod_ids_raw: # <- папка пустая
                logger.warning("No products found!")
                #return []
            # 3. последняя попытка - ищу файл xls, который получаю из экпотра в кабинете affiliate panel        

            prod_ids_raw = campaign_parser.prepare_products_for_campaign_from_xls()
            if not prod_ids_raw: # <- файла или нет или он покоцаный
                logger.error("No products found!")
                return []
  
        prod_ids = self.extract_prod_ids(prod_ids_raw)
        ...
        return campaign_parser._affiliate_product(prod_ids)

    def get_campaign_products(self) -> List[SimpleNamespace]:
        """! Reads product data from JSON files prepared in `prepare_products_for_campaign()`
        @return: A list of campaign products
        """
        logger.info(f"Start processing {self.campaign_category_path}")
        product_files_dir = self.campaign_category_path / self.language / 'parsed_products'
        if not product_files_dir:
            self.prepare_products_for_campaign() 
            
        return j_loads_ns(product_files_dir)

    def extract_prod_ids(self, urls: List[str]) -> List[str]:
        """! Extracts item IDs from a list of URLs.
        @param urls: List of URLs containing item IDs.
        @return: List of extracted item IDs.
        """
        # Регулярное выражение для поиска идентификаторов товаров
        pattern = re.compile(r'(?:item/|/)?(\d+)\.html')
    
        # Извлечение идентификаторов товаров из URL
        logger.info(f"Found {len(urls)} products")
        return [str(match.group(1)) for url in urls if (match := pattern.search(url))]
```

```python
def get_prod_ids(path_to_products):
    product_filenames = get_filenames(path_to_products)
    #print(product_filenames)
    prod_ids = [s.split('.', 1)[0].strip() if '.' in s else s for s in product_filenames]
    return prod_ids
```
